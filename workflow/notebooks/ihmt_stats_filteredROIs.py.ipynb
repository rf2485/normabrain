```python
import pandas as pd
import numpy as np
import json
```

```python
df = pd.read_pickle(snakemake.input[0])
df
```

```python
region_list = ['Left-Cerebral-White-Matter', 'Right-Cerebral-White-Matter', 
               'frontal_left_wm', 'frontal_right_wm', 
               'temporal_left_wm', 'temporal_right_wm',
               'parietal_left_wm', 'parietal_right_wm',
               'occipital_left_wm', 'occipital_right_wm']
df = df.query('region in @region_list')
df
```

```python
def hist_to_json(x, bins=20):
    counts, edges = np.histogram(x, bins=bins)
    return json.dumps([
        {
            "bin_start": float(start),
            "bin_end": float(end),
            "count": int(count),
        }
        for count, start, end in zip(counts, edges[:-1], edges[1:])
    ])

df["hist"] = df["data"].apply(hist_to_json)
```

```python
df = df.drop("data", axis=1)
df
```

```python
df.to_csv(snakemake.output[0])
```
